```python
import numpy as np
import pandas as pd
import itertools
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import PassiveAggressiveClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
df = pd.read_csv('news.zip')
```

```python
df.shape
```

```python
df.head()
```

```python
labels = df.label
labels.head()
```

```python
x_train,x_test,y_train,y_test = train_test_split(df["text"], labels, test_size = 0.2, random_state = 7)
```

```python
tfidf_vectorizer = TfidfVectorizer(stop_words = 'english', max_df = 0.7)
```

```python
tfid_train = tfidf_vectorizer.fit_transform(x_train)
tfid_test = tfidf_vectorizer.transform(x_test)
```

```python
pac = PassiveAggressiveClassifier(max_iter = 50)
pac.fit(tfid_train, y_train)

y_pred = pac.predict(tfid_test)
score = accuracy_score(y_test, y_pred)
print(f'Accuracy: {round(score*100,2)}%')
```

```python
confusion_matrix(y_test, y_pred, labels=['FAKE', 'REAL'])
```

```python
import pickle

# Save the vectorizer
with open('tfidf_vectorizer.pkl', 'wb') as f:
    pickle.dump(tfidf_vectorizer, f)

# Save the trained model
with open('pac_model.pkl', 'wb') as f:
    pickle.dump(pac, f)

print("Model and vectorizer saved successfully!")
```
